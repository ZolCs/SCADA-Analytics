```python
# Import libraries

import sqlite3
```

```python
# Configuration

sqliteConnection = sqlite3.connect('SCADA_Configuration.db')
cursor = sqliteConnection.cursor()
```

```python
# Create table: Configuration

cursor.execute("DROP TABLE IF EXISTS configuration")

table_creation_query = """
    CREATE TABLE configuration (
        ID INTEGER PRIMARY KEY NOT NULL,
        comment TEXT,
        created TEXT,
        temperature_warning INTEGER,
        temperature_critical INTEGER,
        pressure_warning REAL,
        pressure_critical REAL
    );
"""
cursor.execute(table_creation_query)
```

```python
# Create table: Current_configuration

cursor.execute("DROP TABLE IF EXISTS current_configuration")

table_creation_query = """
    CREATE TABLE current_configuration (
        configuration_id INTEGER
    );
"""
cursor.execute(table_creation_query)
```

```python
sqliteConnection.commit()
```

```python
import sqlite3

conn = sqlite3.connect("SCADA_Configuration.db")
cursor = conn.cursor()

cursor.execute("SELECT * FROM current_configuration")

print(cursor.fetchall())

conn.close()
```

## Module Summary

This module creates the SQLite database structure required by the SCADA configuration system.

The following tables are created:
- `configuration` – stores configuration versions and threshold values.
- `current_configuration` – stores the ID of the currently active configuration.

> **Note:** This module is intended for initial database setup and should only be executed when creating or resetting the configuration database.
