```python
# Data entry and validation

comment_error_message=f"The text can be a maximum of 50 characters long."
ok_message_1=f"The data entry was successful."

while True:
    txt_2 = input('Add a comment:')
    comment= txt_2.strip()
    comment_lenght=len(comment)
    if not (comment_lenght <= 50):   # Hosszúság ellenőrzése
        print(comment_error_message)
        continue
    print(ok_message_1)

    break

temperature_warning_error_message=(
    f"The data is required!\n"
    f"The data can only be an integer!\n"
    f"The warning temperature must be between 100 and 140 °C!"
    )

while True:
    txt_3= input('Set the temperature warning (in °C):')
    raw_temperature_warning=txt_3.strip()  
    if not raw_temperature_warning:   #Kitöltés ellenőrzése
        print(temperature_warning_error_message)
        continue
  
    try:    # Szám ellenőrzés
        temperature_warning = int(raw_temperature_warning)
    except ValueError:
        print(temperature_warning_error_message)
        continue
  
    if not (100 <= temperature_warning <= 140): # Tatomány vizsgálat
        print(temperature_warning_error_message)
        continue
   
    print(ok_message_1)

    break

temperature_critical_error_message=(
    f"The data is required!\n"
    f"The data can only be an integer!\n"
    f"The critical temperature must be between 100 and 140 °C!\n"
    f"Temperature critical must be greater than temperature warning"
    )

while True:
    txt_4= input('Set the temperature critical (in °C):')
    raw_temperature_critical=txt_4.strip()  
    if not raw_temperature_critical:   #Kitöltés ellenőrzése
        print(temperature_critical_error_message)
        continue
  
    try:    # Szám ellenőrzés
        temperature_critical = int(raw_temperature_critical)
    except ValueError:
        print(temperature_critical_error_message)
        continue
  
    if not (100 <= temperature_critical <= 140): # Tatomány vizsgálat
        print(temperature_critical_error_message)
        continue

    if not (temperature_warning < temperature_critical): # A feltétel
        print(temperature_critical_error_message)
        continue
   
    print(ok_message_1)

    break

pressure_warning_error_message=(
    f"The data is required!\n"
    f"The data can only be a float!\n"
    f"The warning pressure must be between 3 and 5 bar!\n"
    )

while True:
    txt_5= input('Set the pressure warning (in bar):')
    raw_pressure_warning=txt_5.strip()  
    if not raw_pressure_warning:   #Kitöltés ellenőrzése
        print(pressure_warning_error_message)
        continue
  
    try:    # Szám ellenőrzés
        pressure_warning = float(raw_pressure_warning)
    except ValueError:
        print(pressure_warning_error_message)
        continue
  
    if not (3 <= pressure_warning <= 5): # Tatomány vizsgálat
        print(pressure_warning_error_message)
        continue

    print(ok_message_1)

    break

pressure_critical_error_message=(
    f"The data is required!\n"
    f"The data can only be a float!\n"
    f"The warning pressure must be between 3 and 5 bar!\n"
    f"pressure critical must be greater than pressure warning"
    )

while True:
    txt_6 = input('Set the pressure critical (in bar):')
    raw_pressure_critical=txt_6.strip()  
    if not raw_pressure_critical:   #Kitöltés ellenőrzése
        print(pressure_critical_error_message)
        continue
  
    try:    # Szám ellenőrzés
        pressure_critical = float(raw_pressure_critical)
    except ValueError:
        print(pressure_critical_error_message)
        continue
  
    if not (3 <= pressure_critical <= 5): # Tatomány vizsgálat
        print(pressure_critical_error_message)
        continue

    if not (pressure_warning < pressure_critical): # A feltétel
        print(pressure_critical_error_message)
        continue

    print(ok_message_1)

    break
```

```python
# Print the solution

text_1= f"SCADA Configuration\n"
text_2= (
    f"Comment: {comment}\n"
)
text_3= (
    f"Temperature Warning: {temperature_warning}\n"
    f"Temperature Critical: {temperature_critical}\n"
    f"Pressure Warning: {pressure_warning}\n"
    f"Pressure Critical: {pressure_critical}\n"
)
text_4= (
    f"- - - - - - - - - - - - - - - \n"
    f"\n"
    f"Configuration Summary\n"
)
text_5=(
    f"Accept?\n"
    f"\n"
    f"Y/N"
)

print(text_1,text_2,text_4,text_3, text_5, sep="\n")
```

```python

option_error= f"Only Y or N can be entered."

while True:
    raw_option = input('Please choose an option (Y/N)')
    option= raw_option.strip()
    if not (option == "Y" or option == "N"):   # Control
        print(option_error)
        continue
    answer_message=f"Answer accepted. Value: {option}"    
    print (answer_message)
    
    break
```

```python
from datetime import datetime

now=datetime.now()
timestamp=now.strftime("%Y-%m-%d %H:%M:%S")

print(timestamp)
```

```python
# Configuration

sqliteConnection = sqlite3.connect('SCADA_Configuration.db')
cursor = sqliteConnection.cursor()
```

```python
# Add data to the configuration table
if option=="Y":
    cursor.execute('''
    INSERT INTO configuration (comment, created, temperature_warning, temperature_critical, pressure_warning, pressure_critical)
    VALUES (?, ?, ?, ?, ?, ?)
    ''', (comment, timestamp, temperature_warning, temperature_critical, pressure_warning, pressure_critical)) 
```

```python
# Select the last ID
selected_ID = cursor.lastrowid
print(selected_ID)
```

```python
# Add the last ID to current_configuration table
cursor.execute('DELETE FROM current_configuration')
cursor.execute('''
        INSERT INTO current_configuration (configuration_id)
        VALUES (?)
        ''', (selected_ID,))
sqliteConnection.commit()
sqliteConnection.close()
```

## Module Summary

This module handles the creation of a new SCADA configuration.

User input is validated before being stored in the `configuration` table. After a configuration is successfully created, its ID is written to `current_configuration`, replacing the previously active configuration.

This ensures that the system always has a clearly defined current configuration while preserving previous configuration versions.

### Next Step
The active configuration can be read by the data generator and other modules that require the current SCADA threshold values.
